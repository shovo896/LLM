```python
def parse_angles(value):
    """
    Convert angle representation into list of floats.
    """

    if isinstance(value, (list, tuple, np.ndarray)):
        vals = value

    elif pd.isna(value):
        return []

    elif isinstance(value, str):

        text = value.strip()

        try:
            vals = ast.literal_eval(text)
        except:
            # fallback for simple comma-separated values
            text = text.replace("[", "").replace("]", "")
            try:
                vals = [float(x) for x in text.split(",")]
            except:
                return []

    else:
        vals = [value]

    if np.isscalar(vals):
        vals = [vals]

    output = []

    for x in vals:
        try:
            x = float(x)

            if -60 <= x <= 60:
                output.append(x)

        except:
            pass

    return sorted(output)
```

```python
import re

def parse_angles(value):
    """
    Parse examples like:
    '[ 6 -60]'
    '[28]'
    '[-17 -52 -22 -35 19]'
    """

    if pd.isna(value):
        return []

    text = str(value)

    numbers = re.findall(
        r'[-+]?\d*\.?\d+',
        text
    )

    angles = [
        float(x)
        for x in numbers
    ]

    angles = [
        x for x in angles
        if -60 <= x <= 60
    ]

    return angles
```

```python


import torch
from torch.utils.data import Dataset, DataLoader

class RadarRowDataset(Dataset):
    def __init__(
        self,
        X,
        y_angle,
        y_count,
        augment=False
    ):
        self.X = torch.tensor(
            X,
            dtype=torch.float32
        )

        self.y_angle = torch.tensor(
            y_angle,
            dtype=torch.float32
        )

        self.y_count = torch.tensor(
            y_count,
            dtype=torch.long
        )

        self.augment = augment

    def __len__(self):
        return len(self.X)

    def __getitem__(self, idx):

        x = self.X[idx].clone()

        if self.augment:
            # small additive noise
            if torch.rand(1).item() < 0.3:
                x = x + 0.01 * torch.randn_like(x)

            # random antenna dropout
            if torch.rand(1).item() < 0.2:
                mask = torch.rand_like(x) < 0.05
                x[mask] = 0.0

        return (
            x,
            self.y_angle[idx],
            self.y_count[idx]
        )
```

```python


class RadarMamba3(nn.Module):

    def __init__(
        self,
        n_antennas=29,
        d_model=128,
        d_state=64,
        n_layers=3,
        n_angle_bins=121,
        max_targets=5,
        mimo_rank=4
    ):
        super().__init__()

        self.n_antennas = n_antennas

        # Each antenna scalar -> d_model embedding
        self.input_embedding = nn.Sequential(
            nn.Linear(1, d_model),
            nn.GELU(),
            nn.LayerNorm(d_model)
        )

        # learnable antenna positional embedding
        self.antenna_embedding = nn.Parameter(
            torch.randn(
                1,
                n_antennas,
                d_model
            ) * 0.02
        )

        self.blocks = nn.ModuleList([
            RadarMambaBlock(
                d_model=d_model,
                d_state=d_state,
                headdim=32,
                mimo_rank=mimo_rank,
                chunk_size=16
            )
            for _ in range(n_layers)
        ])

        self.final_norm = nn.LayerNorm(d_model)

        self.feature_head = nn.Sequential(
            nn.Linear(d_model, d_model),
            nn.GELU(),
            nn.Dropout(0.2)
        )

        self.angle_head = nn.Linear(
            d_model,
            n_angle_bins
        )

        self.count_head = nn.Linear(
            d_model,
            max_targets
        )

    def forward(self, x):

        # x: [B, 29]

        x = x.unsqueeze(-1)

        # [B,29,1]

        x = self.input_embedding(x)

        # [B,29,d_model]

        x = x + self.antenna_embedding

        for block in self.blocks:
            x = block(x)

        x = self.final_norm(x)

        # pool over antenna sequence
        x_mean = x.mean(dim=1)

        x_max = x.max(dim=1).values

        x = 0.5 * (x_mean + x_max)

        x = self.feature_head(x)

        angle_logits = self.angle_head(x)

        count_logits = self.count_head(x)

        return angle_logits, count_logits
```

```python


from scipy.optimize import linear_sum_assignment

def angular_set_mae(
    true_angles,
    pred_angles
):

    true_angles = np.array(
        true_angles,
        dtype=float
    )

    pred_angles = np.array(
        pred_angles,
        dtype=float
    )

    if len(true_angles) == 0 or len(pred_angles) == 0:
        return 120.0

    cost = np.abs(
        true_angles[:, None]
        - pred_angles[None, :]
    )

    row_ind, col_ind = linear_sum_assignment(
        cost
    )

    matched_error = cost[
        row_ind,
        col_ind
    ].mean()

    count_penalty = (
        abs(
            len(true_angles)
            - len(pred_angles)
        )
        * 10.0
    )

    return matched_error + count_penalty
```

```python


@torch.no_grad()
def evaluate_model(
    model,
    loader,
    true_angle_lists
):

    model.eval()

    total_loss = 0

    correct_count = 0
    total_samples = 0

    all_predictions = []

    for X_batch, y_angle, y_count in loader:

        X_batch = X_batch.to(device)
        y_angle = y_angle.to(device)
        y_count = y_count.to(device)

        with torch.autocast(
            device_type="cuda",
            dtype=torch.float16,
            enabled=use_amp
        ):

            angle_logits, count_logits = model(
                X_batch
            )

            loss_angle = angle_criterion(
                angle_logits,
                y_angle
            )

            loss_count = count_criterion(
                count_logits,
                y_count
            )

            loss = (
                loss_angle
                +
                COUNT_LOSS_WEIGHT
                * loss_count
            )

        total_loss += loss.item()

        pred_count = (
            count_logits.argmax(dim=1)
        )

        correct_count += (
            pred_count == y_count
        ).sum().item()

        total_samples += len(
            X_batch
        )

        preds = decode_predictions(
            angle_logits,
            count_logits
        )

        all_predictions.extend(
            preds
        )

    mae_list = []

    for true_a, pred_a in zip(
        true_angle_lists,
        all_predictions
    ):

        mae_list.append(
            angular_set_mae(
                true_a,
                pred_a
            )
        )

    return {
        "loss":
            total_loss / len(loader),

        "count_accuracy":
            correct_count /
            total_samples,

        "angle_mae":
            np.mean(mae_list),

        "predictions":
            all_predictions
    }
```
