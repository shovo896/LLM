```python
import kaggle_benchmarks as kbench

@kbench.task(
    name="machine_learning_basics",
    description="Tests whether an LLM understands basic machine learning concepts."
)
def machine_learning_basics(llm):

    response = llm.prompt(
        """
        Answer this question briefly:

        What is overfitting in machine learning,
        and mention one method to reduce it?
        """
    )

    response_lower = response.lower()

    # Model should mention overfitting-related concepts
    kbench.assertions.assert_true(
        (
            "training" in response_lower
            or "train" in response_lower
        ),
        expectation="The answer should discuss training performance."
    )

    # Model should mention at least one common solution
    solutions = [
        "dropout",
        "regularization",
        "early stopping",
        "data augmentation",
        "cross-validation"
    ]

    found_solution = any(x in response_lower for x in solutions)

    kbench.assertions.assert_true(
        found_solution,
        expectation="The answer should mention at least one method for reducing overfitting."
    )


machine_learning_basics.run(kbench.llm)
```
