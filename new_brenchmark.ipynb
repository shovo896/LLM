```python
import kaggle_benchmarks as kbench

@kbench.task(name="calculate classification accuracy from confusion matrix")
def calculate_classification_accuracy_from_confusion_matrix() -> float:
    tp = 80
    tn = 90
    fp = 10
    fn = 20

    total_predictions = tp + tn + fp + fn
    correct_predictions = tp + tn

    accuracy = (correct_predictions / total_predictions) * 100
    rounded_accuracy = round(accuracy, 2)

    kbench.assertions.assert_equal(
        85.00,
        rounded_accuracy,
        expectation="The calculated accuracy should be exactly 85.00%."
    )

    return rounded_accuracy

calculate_classification_accuracy_from_confusion_matrix.run()
```
